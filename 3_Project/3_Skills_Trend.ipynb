# How are in-demand skills trending for Data Analysts?

#### Methology
1. Aggregate skill counts monthly
2. Re-analyze based on percentage of total jobs
3. Plot the monthly skill demand

```python
# Importing Libraries
import ast
import pandas as pd
import seaborn as sns
from datasets import load_dataset
import matplotlib.pyplot as plt 

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

## 1. Aggregate skill counts monthly

```python
# Filter data down to Data Analyst in the US
df_DA_US = df[(df['job_title'] == 'Data Analyst') & (df['job_country'] == 'United States')].copy()
df_DA_US
```

```python
# Derive the Job Posted Month Number and store it in a new column
df_DA_US['job_posted_month_no'] = df_DA_US['job_posted_date'].dt.month
df_DA_US
```

```python
# Expand the job skills column into multiple rows per job title
df_DA_US_explode = df_DA_US.explode('job_skills')
df_DA_US_explode
```

```python
# Pivot the data so we get the count of skills for each month number the jobs were posted
# Fill any NaN values(where there are no records for that skill and month combo) with 0
df_DA_US_pivot = df_DA_US_explode.pivot_table(index='job_posted_month_no', columns='job_skills', aggfunc='size', fill_value=0)
df_DA_US_pivot
```

```python
# Create a Total row to calculate the counts for the entire year
df_DA_US_pivot.loc['Total'] = df_DA_US_pivot.sum()
df_DA_US_pivot
```

```python
# Sort by Total Row
df_DA_US_pivot = df_DA_US_pivot[df_DA_US_pivot.loc['Total'].sort_values(ascending=False).index]
df_DA_US_pivot
```

```python
# Drop the Total Row now that the skills are sorted in the proper order
df_DA_US_pivot = df_DA_US_pivot.drop('Total')
df_DA_US_pivot
```

## 2. Re-analyze based on percentage of total jobs

```python
# Calculate Total Skills for all job postings per month
DA_totals = df_DA_US.groupby('job_posted_month_no').size()
DA_totals
```

```python
# Calculate the percentages for each skill by dividing the skill count for the month by the total skills for that month
# Calculation should be done for each month(axis=0 where 0 is the index column or the 'job_posted_month_no')
# Divide the totals by 100 to get the values in percentage form
df_DA_US_percent = df_DA_US_pivot.div(DA_totals/100, axis=0)
df_DA_US_percent
```

```python
# Change the month number to month name
df_DA_US_percent = df_DA_US_percent.reset_index()
df_DA_US_percent['job_posted_month'] = df_DA_US_percent['job_posted_month_no'].apply(lambda x: pd.to_datetime(x, format='%m').strftime('%b'))
df_DA_US_percent = df_DA_US_percent.set_index('job_posted_month')
df_DA_US_percent = df_DA_US_percent.drop(columns='job_posted_month_no')
df_DA_US_percent
```

## 3. Plot the monthly skill demand

```python
# Grab only the top 5 skills
df_plot = df_DA_US_percent.iloc[:, :5]
df_plot
```

```python
# Plot the Top 5 Skills
num_skills = len(df_plot.columns)
num_months = len(df_plot)
lbl_buffer = .2

sns.lineplot(data=df_plot, dashes=False, palette='tab10')
sns.set_theme(style='ticks')
sns.despine() # remove the top and right border

plt.title('Trending Top Skills for Data Analysts in the US')
plt.ylabel('Likelihood in Job Posting')
plt.xlabel('2023')
plt.legend().remove()

# show percentage on Y axis labels
from matplotlib.ticker import PercentFormatter
ax = plt.gca()
ax.yaxis.set_major_formatter(PercentFormatter(decimals=0))

# Label each line with the skill name
for i in range(num_skills):
    plt.text((num_months - 1) + lbl_buffer, df_plot.iloc[-1, i], df_plot.columns[i])

plt.show()
```
